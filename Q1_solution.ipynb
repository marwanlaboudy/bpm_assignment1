 Setup: Ne2ra el file

```python
import csv
import re
from datetime import datetime

file_path = '/content/Q1_raw_event_log.csv'

with open(file_path, 'r', encoding='utf-8') as f:
    reader = csv.reader(f)
    header = next(reader)
    rows = list(reader)
```

### Q1.1

```python
# 3adad el rows mn gher el header
q1_1 = len(rows)
print(f"Q1.1: {q1_1}")
```

### Q1.2

```python
# 3adad el unique case_ids el msh fadya
unique_cases = set(r[0].strip() for r in rows if r[0].strip() != '')
q1_2 = len(unique_cases)
print(f"Q1.2: {q1_2}")
```

### Q1.3

```python
# rows feha case_id fady [CASE-MIS]
q1_3 = sum(1 for r in rows if r[0].strip() == '')
print(f"Q1.3: {q1_3}")
```

### Q1.4

```python
# rows feha timestamp fady [TS-MIS]
q1_4 = sum(1 for r in rows if r[2].strip() == '')
print(f"Q1.4: {q1_4}")
```

### Q1.5

```python
# rows feha date bas mn gher wa2t YYYY-MM-DD [TS-IMP]
date_only_pattern = re.compile(r'^\d{4}-\d{2}-\d{2}$')
q1_5 = sum(1 for r in rows if date_only_pattern.match(r[2].strip()))
print(f"Q1.5: {q1_5}")
```

### Q1.6

```python
# rows feha resource fady [RES-MIS]
q1_6 = sum(1 for r in rows if r[3].strip() == '')
print(f"Q1.6: {q1_6}")
```

### Q1.7

```python
# duplicate rows metkarara belzabt fe kol el columns [EV-CIL]
seen_rows = set()
duplicates = 0

for r in rows:
    row_tuple = tuple(r)
    if row_tuple in seen_rows:
        duplicates += 1
    else:
        seen_rows.add(row_tuple)

q1_7 = duplicates
print(f"Q1.7: {q1_7}")
```

### Q1.8

```python
# 3adad el activities el unique zay ma heya maktoba
unique_activities = set(r[1] for r in rows)
q1_8 = len(unique_activities)
print(f"Q1.8: {q1_8}")
```

### Q1.9

```python
# rows feha activity msh men el 8 official activities [ACT-IMP]
official_activities = {
    "Submit Application",
    "Verify Documents",
    "Check Eligibility",
    "Assign Advisor",
    "Select Courses",
    "Pay Tuition",
    "Issue Student Card",
    "Confirm Enrollment"
}

q1_9 = sum(1 for r in rows if r[1] not in official_activities)
print(f"Q1.9: {q1_9}")
```

### Q1.10

```python
# cases mafehash "Select Courses" b3d el normalization [EV-MIL]
def clean_label(text):
    return re.sub(r'_', ' ', text.strip().lower())

mandatory_cleaned = clean_label("Select Courses")

cases_with_mandatory = set()
for r in rows:
    cid = r[0].strip()
    act = r[1]
    if cid != '':
        if clean_label(act) == mandatory_cleaned:
            cases_with_mandatory.add(cid)

q1_10 = len(unique_cases - cases_with_mandatory)
print(f"Q1.10: {q1_10}")
```

### Q1.11

```python
# cases feha timestamps msh martaba sa7 chronologically [POS-INC]
case_times = {}

for r in rows:
    cid = r[0].strip()
    ts_str = r[2].strip()
    if cid == '':
        continue
    try:
        dt = datetime.strptime(ts_str, '%Y-%m-%d %H:%M:%S')
        if cid not in case_times:
            case_times[cid] = []
        case_times[cid].append(dt)
    except ValueError:
        pass

out_of_order_cases = 0
for cid, times in case_times.items():
    if not all(times[i] <= times[i+1] for i in range(len(times)-1)):
        out_of_order_cases += 1

q1_11 = out_of_order_cases
print(f"Q1.11: {q1_11}")
```

### Q1.12

```python
# pattern el missing resource: intermittent (INT = 2)
q1_12 = 2
print(f"Q1.12: {q1_12}")
```

### Q1.13

```python
# akbar moshkela fe 3adad el rows/cases (ACT-IMP = 5)
counts = {
    1: ("TS-MIS", q1_4),
    2: ("TS-IMP", q1_5),
    3: ("RES-MIS", q1_6),
    4: ("EV-CIL", q1_7),
    5: ("ACT-IMP", q1_9),
    6: ("CASE-MIS", q1_3),
    7: ("EV-MIL", q1_10),
    8: ("POS-INC", q1_11),
}

q1_13 = max(counts, key=lambda k: counts[k][1])
print(f"Q1.13: {q1_13}")
```
